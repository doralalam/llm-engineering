#### Fine Tuning a Model

- Fine-tuning is an inference time technique, because in this technique, the model's weights/parameters gets updated based on the training data

#### Difference between Training Time Technique and Inference Time Technique:

- **Training Time Technique**

    - Training involves setting the parameters of the base model based on inputs and outputs

    - Example: Fine-Tuning

- **Inference Time Technique**

    - Inference time technique is a trick we apply to improve the performance of the model during it's run-time

    - Example: Prompt Engineering (better prompts), RAG (knowledge retrieval), In-Context learning (examples before asking a question for better output)

#### Important Terminology

- **Generalization**

    - The ability of a model to make good predictions on unseen data is called "Generalization"

- **Transfer Learnining**

    - The process of having a base model and using additional training data to fine-tune it for a custom task is called "Transfer Learning"

#### Note:

- While training or fine-tuning a model on a custom data, it is always a best practice to split the data into 3 datasets:

    1. Training Data (80%)      - Used for training the model

    2. Validation Data (10%)    - Used for validating the model after training data. Alter the parameters incase not satisfied with the performance and repeat the process until the performance is satisfied

    3. Test Data (10%)          - Used for ultimate testing of the model

#### 5 - Step Strategy to Select, Train and Apply an LLM to a Commercial Problem

- Step - 1 : Understand

    - Gather business requirements for the task

    - Indentify performance criteria like model metrics and business metrics

    - Understand data quality, quantity and format

    - Determine non-functionals: cost constraints, scalability, latency, R&D / build budget and implementation timeline

- Step - 2 : Prepare

    - Research existing / non-LLM solutions [Potenital baseline model]

    - Compare relevant LLMs

    - Curate Data: Clean, Pre-process and split

- Step - 3 : Select

    - Choose LLMs

    - Experiment

    - Train & Validate with curated data

- Step - 4 : Customize

    - Techniques to optimize the performance of the model

    - Inferece Time Techniques: Prompting, RAG, Agents (orchestrating multiple agents)

    - Training Time Techniques: Fine-Tuning i.e., teaching a model a new skill that will allow it to generalize (to provide best output for unseen data)

- Step - 5 : Deploy

    - Determine API between model and platforms

    - Identify model hosting and deployment architecture

#### Chinchilla Scaling Law:

- Chinchilla's Scaling Law (published by DeepMind in 2022) is a principle that dictates `how to optimally split a compute budget between model size and dataset size to build the most capable LLM`

- Before chinchilla, AI research focused mainly on making the models as large as possible while keeping the training data relatively small

- Chinchilla proved that most exisiting models was severely under-trained relative to their size and that parameters and training data should actually scale at en equal ratio of 1:1

#### Core Golden Rule: 20 Tokens per parameter

Chinchilla introduced a simple operational rule of thumb for compute-optimal pre-training:

`Optimal Training Tokens (D) ~= 20 X Model Parameters (N)`

- For every 1 billion parameters, a model should be trained on approximately 20 billion dataset tokens.

- If your training compute budget increases by 10×, you should scale both your model size (N) and training data (D) by roughly 3.16× (sqrt(10)), rather than expanding parameter count alone.

#### Modern Impact

- While Chinchilla established the compute-optimal threshold for training efficiency, modern open-weights models (like Meta's Llama series) deliberately break Chinchilla's ratio by training on far more than 20 tokens per parameter:

- `Llama 3 8B was trained on 15 trillion tokens—a ratio of nearly 1,875 tokens per parameter.`

#### Why over-train beyond Chinchilla optimal?

- Chinchilla optimizes for lowest training cost. However, in production, 

- `a model is trained once but queried billions of times`. 

- Over-training a smaller 8B model costs more during pre-training, but drastically cuts hosting costs and response latency for millions of downstream users compared to serving a larger 70B model.

#### Difference between Pruning and Quantizing

1. Pruning:

    - Reducing the number of parameters (or) weights of the model to save the storage and compute cost

        - Deep learning models are famously over-parameterized—many weights are near zero and contribute almost nothing to the outputs.

        - Unstructured pruning can often remove 20% to 50%+ of weights with virtually zero loss in accuracy/performance, provided the remaining model is fine-tuned afterward.
        
        - The real catch with Pruning: While it reduces the theoretical number of operations (FLOPs), unstructured pruning (removing individual scattered weights) creates sparse matrices. Standard GPUs are built for dense matrix math, so unstructured pruning often does not speed up real-world inference unless you use specialized hardware/software engines (like TensorRT) or use structured pruning (removing entire layers or channels).

2. Quantizing:

    - Reducing the size of each parameter i.e., from 32-bit float values to 4-bit values to save the storage and compute cost

        - Quantization is widely preferred in industry, because it gives guaranteed, instant hardware speedups and VRAM reductions.

        - Modern GPUs have dedicated hardware (like Tensor Cores) optimized specifically for 8-bit (INT8) and 4-bit integer operations.

        - Quantization keeps the network structure dense, meaning memory layout stays identical, memory bandwidth demands drop by up to 4x–8x, and speedups happen out of the box without needing specialized sparse-matrix engines.

#### Limitations of Benchmarks:

1. Training data contamination - Leakage of test set questions during the training stage might make the model overfit for those questions but fail for general test

2. Not consistently applied

3. Too narrow in scope like GPQA 

4. Hard to measure nuanced reasoning

5. Saturation

6. Overfitting

7. Frontier LLMs may be aware that they are being evaluated

#### Training Cut-off time

- A training cut-off time (or knowledge cutoff) is the date up to which an AI model was supplied with data during its training process.

#### How AI Handles Information Past Its Cut-Off

When you ask an AI about something that happened after its training cut-off, it typically relies on one of two approaches:

- Base Knowledge (No Search):	The model relies strictly on its internal memory. It will either state that it doesn't know or inform you of its cut-off date.	
    
    - Asking for last night's sports scores without web access.

- Web Search & Retrieval (RAG):	The AI uses tools to search the live internet in real time, grounding its response in current sources before answering.	

    - Asking for today's weather or news headline.

```python
import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv(override=True)

GEMINI_BASE_URL = os.getenv('GEMINI_BASE_URL')
GEMINI_API_KEY = os.getenv('GEMINI_API_KEY')

if GEMINI_API_KEY:
    print(f'GEMINI API KEY exists and starts with {GEMINI_API_KEY[0:3]}')
else:
    print('GEMINI API KEY does not exist')

gemini = OpenAI(base_url=GEMINI_BASE_URL, api_key=GEMINI_API_KEY)
```

```python
system_message = "You are a helpful assistant!"

def message_gemini(prompt):
    response = gemini.chat.completions.create(
        model = 'gemini-3.6-flash',
        messages = [
            {'role':'system', 'content':system_message},
            {'role':'user', 'content':prompt}
        ]
    )
    return response.choices[0].message.content
```

```python
message_gemini("What is your training cut-off time?")
```

```python
message_gemini("What is the today's date?")
```

#### Gradio

- Gradio is an open source Python framework designed to let you build, customize and share interactive web interfaces for machine learning models, APIs or Python functions

- Instead of building a frontend web application using HTML, CSS, JavaScript, Gradio lets you write Python code to launch a fully functioning, interactive web UI in minutes

```python
%pip install gradio
```

```python
import gradio as gr
```

```python
gr.Interface(fn=message_gemini, inputs="textbox", outputs="textbox", flagging_mode="never").launch()
```

```python
''' 
gr.Interface(...) ------    creates the UI
fn=message_gemini ------    tells the Gradio which function to call
flagging_mode = "never"     disables saving example submissions
.launch() --------------    starts the local web app in your browser    

NOTE: 

flagging_mode controls whether Gradio should allow users to mark or save example outputs as “flagged” for review.

If the user thinks the answer is wrong or poor, they can click a “flag” button

In practice, it is mainly used for:

    - collecting feedback on model outputs
    - labeling bad or unsafe responses
    - building datasets for evaluation or testing

Common values:

    "never" → disables flagging completely
    "manual" → lets users click a flag button to mark outputs
    "auto" → automatically flags outputs based on certain conditions
'''
```

#### Gradio's Share Tool

```python
gr.Interface(fn=message_gemini, inputs="textbox", outputs="textbox", flagging_mode="never").launch(share=True)

# Adding share=True means that it can be accessed publicly
# Share link is temporary and will last for upto 1 week (best effort)
# For free permanent hosting, we can use HuggingFace spaces
```

```python
gr.Interface(fn=message_gemini, inputs="textbox", outputs="textbox", flagging_mode="never").launch(inbrowser=True)

# Adding inbowser=True, opens a new browser window automatically
```

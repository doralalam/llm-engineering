#### Training Cut-off time

- A training cut-off time (or knowledge cutoff) is the date up to which an AI model was supplied with data during its training process.

#### How AI Handles Information Past Its Cut-Off

When you ask an AI about something that happened after its training cut-off, it typically relies on one of two approaches:

- Base Knowledge (No Search):	The model relies strictly on its internal memory. It will either state that it doesn't know or inform you of its cut-off date.	
    
    - Asking for last night's sports scores without web access.

- Web Search & Retrieval (RAG):	The AI uses tools to search the live internet in real time, grounding its response in current sources before answering.	

    - Asking for today's weather or news headline.

```python
import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv(override=True)

GEMINI_BASE_URL = os.getenv('GEMINI_BASE_URL')
GEMINI_API_KEY = os.getenv('GEMINI_API_KEY')

if GEMINI_API_KEY:
    print(f'GEMINI API KEY exists and starts with {GEMINI_API_KEY[0:3]}')
else:
    print('GEMINI API KEY does not exist')

gemini = OpenAI(base_url=GEMINI_BASE_URL, api_key=GEMINI_API_KEY)
```

```python
SYSTEM_MESSAGE = "You are a helpful assistant!"

def message_gemini(prompt):
    response = gemini.chat.completions.create(
        model = 'gemini-3.6-flash',
        messages = [
            {'role':'system', 'content':SYSTEM_MESSAGE},
            {'role':'user', 'content':prompt}
        ]
    )
    return response.choices[0].message.content
```

```python
message_gemini("What is your training cut-off time?")
```

```python
message_gemini("What is the today's date?")
```

#### Streaming the Response

For streaming, we need to use yield instead of return

```python
def stream_gemini(prompt):
    messages = [
        {'role':'system', 'content':SYSTEM_MESSAGE},
        {'role':'user', 'content':prompt}
    ]

    stream = gemini.chat.completions.create(
        model='gemini-3.6-flash',
        messages=messages,
        stream=True
    )

    response=""
    for chunk in stream:
        response += chunk.choices[0].delta.content or ""
        yield response
```

```python
input_message = gr.Textbox(label = 'Your message', info="Enter a message for gemini-3.6-flash", lines=2)
output_message = gr.Markdown(label='Response: ')

gr.Interface(
    fn=stream_gemini,
    title='GEMINI Streaming',
    inputs=[input_message],
    outputs=[output_message],
    examples= [
        'Explain about Tokenization in 5 lines', 
        'What is Back-Propagation?'
        ],
    flagging_mode="never"
).launch()
```

#### Difference between return and yield

return:

    - ends the function completely
    - sends one value back
    - the function can not continue after it

    - Example: 
        def f():
            return "hello"

    - When called this function, it gives "hello" and stops

yield:

    - pauses the function and saves its state
    - returns one value, but the function can resume later
    - used in generators
    - must access the values in yield using looping which is known as "Generators"

    - Example:
        def f():
            yield "hello"
            yield "world"
            yield "!!!"

    - When the function is called, it does not finish the function.
    - It returns the first value, then resumes from where it left off on the next call

Main Difference:

    - return    - one final result and done
    - yield     - a stream of results over time
        



```python
# Demonstration of the difference between return and yield

def returning():
    return "Hello"

def yielding():
    yield "hello"
    yield "world"
    yield "!!!"
```

```python
# yield using Generators, otherwise, it gives the objects as output

for item in yielding():
    print(item)
```

```python
GROQ_BASE_URL = os.getenv('GROQ_BASE_URL')
GROQ_API_KEY = os.getenv('GROQ_API_KEY')

gpt = OpenAI(base_url=GROQ_BASE_URL, api_key=GROQ_API_KEY)

def stream_gpt(prompt):
    messages = [
        {'role':'system', 'content':SYSTEM_MESSAGE},
        {'role':'user', 'content':prompt}
    ]

    stream = gpt.chat.completions.create(
        model = "openai/gpt-oss-120b",
        messages=messages,
        stream=True
    )

    response = ""
    for chunk in stream:
        response += chunk.choices[0].delta.content or ""
        yield response
```

```python
message_input = gr.Textbox(label="Your message:", info="Enter a message for gpt-oss-120b", lines=7)
message_output = gr.Markdown(label="Response:")

view = gr.Interface(
    fn=stream_gpt,
    title="GPT", 
    inputs=[message_input], 
    outputs=[message_output], 
    examples=[
        "Explain the Transformer architecture to a layperson",
        "Explain the Transformer architecture to an aspiring AI engineer",
        ], 
    flagging_mode="never"
    )
view.launch()
```

#### Dropdown

```python
def select_model(prompt, model):
    if model=='GEMINI':
        result = stream_gemini(prompt)
    elif model == 'GPT':
        result = stream_gpt(prompt)
    else:
         raise ValueError('Unknown Model')
    yield from result           # since we are handling the generators for streaming

    ''' 
    Since we are streaming [both stream_gpt() and stream_gemini() are generators], we use yield from result
    This is similar to 

    for item in result:
        yield item

    This means pass through everything produced by the generator
    
    Note:
    It is useful when you want one function to forward the streaming output of another function
    '''
```

```python
input_message = gr.Textbox(label='Your message', info='Enter a message for the LLM', lines=2)
input_model = gr.Dropdown(['GEMINI', 'GPT'], label='Select model', value='GPT')
output_message = gr.Markdown(label='Response: ')

gr.Interface(
    fn=select_model,
    title = 'LLM',
    inputs=[input_message, input_model],
    outputs=[output_message],
    examples= [
            ["Explain the Transformer architecture to a layperson", "GPT"],
            ["Explain the Transformer architecture to an aspiring AI engineer", "GEMINI"]
        ], 
    flagging_mode="never"
).launch()
```

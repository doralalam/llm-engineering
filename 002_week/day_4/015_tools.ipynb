### Tools or Function Calling

- Tools also called as Function Calling refer to external functions, APIs, databases or external environments that an LLM can trigger to overcome its inherent limitations

- While an LLM excels at language understanding, reasoning and pattern recognition, it operates as static black box after the training

- Giving an LLM access to a tool converst it from a passive text generator into an active agent capable of taking actions in the real world



#### Why LLMs need tools?

1. Knowledge Cut-offs:

    - They can't know real time or private information (example: current stocks prices, internal company documents)

2. Deterministic Computation Deficits:

    - They struggle with exact math, long calculations, precise spatial reasoning and complex logic exectution

3. Lack of Side effects:

    - They can't perform actions on external systems (example: sending an email, querying an external database etc.,)

#### Solution

Tools bridge those gaps by offloading specific execution tasks to purpose-built external programs

#### The Tool Cycle

##### How Tool usage works

- An LLM doesn't run Python code or query a database directly in it's transformer layers

- Instead, it generates structured text (usually JSON) that instructs an external runner to execute a function on its behalf

    ┌───────────────────────────────────────────────────────────┐
    │                    1. System & User Prompt                │
    │ User: "What is 1,234 * 5,678 and current weather in NY?"  │
    │ System: Prompt includes tool definitions (schema/JSON)   │
    └─────────────────────────────┬─────────────────────────────┘
                                │
                                ▼
    ┌───────────────────────────────────────────────────────────┐
    │                     2. LLM Tool Choice                    │
    │ Model generates structured JSON call:                     │
    │   - Function: `calculator(a=1234, b=5678, op='mul')`       │
    │   - Function: `get_weather(location='New York')`          │
    └─────────────────────────────┬─────────────────────────────┘
                                │
                                ▼
    ┌───────────────────────────────────────────────────────────┐
    │                3. External Execution (Client)             │
    │ Host runtime parses JSON, calls real APIs/code,           │
    │ gets raw outputs: `7006652` and `"22°C, Sunny"`.          │
    └─────────────────────────────┬─────────────────────────────┘
                                │
                                ▼
    ┌───────────────────────────────────────────────────────────┐
    │                    4. Final LLM Response                  │
    │ Client sends tool results back into model context.        │
    │ Model returns: "1,234 * 5,678 is 7,006,652, and the      │
    │ weather in New York is currently 22°C and sunny."         │
    └───────────────────────────────────────────────────────────┘

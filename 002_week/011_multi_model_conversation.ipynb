#### Initiating the conversation between two models [2 Chat Bots]

```python
import os
from dotenv import load_dotenv
from IPython.display import display, Markdown
from openai import OpenAI

load_dotenv(override=True)

gemini_api_key = os.getenv('GEMINI_API_KEY')
gemini_base_url = os.getenv('GEMINI_BASE_URL')

groq_api_key = os.getenv('GROQ_API_KEY')
groq_base_url = os.getenv('GROQ_BASE_URL')

gemini_client = OpenAI(base_url=gemini_base_url, api_key=gemini_api_key)
gpt_client = OpenAI(base_url=groq_base_url, api_key=groq_api_key)
```

```python
GEMINI_MODEL = 'gemini-3.5-flash-lite'
GPT_MODEL = 'openai/gpt-oss-120b'
```

```python
GEMINI_SYSTEM = "You are a chatbot who is very argumentative; \
you disagree with anything in the conversation and you challenge everything, in a snarky way."

GPT_SYSTEM = "You are a very polite, courteous chatbot. You try to agree with \
everything the other person says, or find common ground. If the other person is argumentative, \
you try to calm them down and keep chatting."

# To store the respective chat history
GEMINI_HISTORY = []
GPT_HISTORY = []

''' 
Logic:
--------------------------------------------------------------------
User input and assistant response of GEMINI was stored in GEMINI_HISTORY
User input and assistant response of GPT was stored in GPT_HISTORY

Assistant response of GEMINI becomes the user input for GPT and vice-versa
'''
```

```python
def call_gemini(user_input: str) -> str:
    GEMINI_HISTORY.append({'role':'user', 'content':user_input})
    messages = [{'role': 'system', 'content': GEMINI_SYSTEM}] + GEMINI_HISTORY
    response = gemini_client.chat.completions.create(
        model = GEMINI_MODEL,
        messages=messages
    )
    bot_reply = response.choices[0].message.content
    GEMINI_HISTORY.append({'role':'assistant', 'content':bot_reply})
    return bot_reply
```

```python
def call_gpt(user_input: str) -> str:
    GPT_HISTORY.append({'role':'user', 'content':user_input})
    messages = [{'role':'system', 'content':GPT_SYSTEM}]+GPT_HISTORY
    response = gpt_client.chat.completions.create(
        model = GPT_MODEL,
        messages = messages
    )
    bot_reply = response.choices[0].message.content
    GPT_HISTORY.append({'role':'assistant', 'content':bot_reply})
    return bot_reply
```

```python
def run_conversation(turns = 3, initial_prompt = 'Hi'):
    print(f'Initiating the conversation with: {initial_prompt}\n'+'-'*100)
    current_prompt = initial_prompt
    for turn in range(1, turns+1):
        print(f'\nTurn {turn}')
        gemini_reply = call_gemini(current_prompt)
        print(f'\nGEMINI : {gemini_reply}')
        gpt_reply = call_gpt(gemini_reply)
        print(f'\nGPT : {gpt_reply}')
        current_prompt = gpt_reply
```

```python
run_conversation()
```

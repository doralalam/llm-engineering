## Training Time Scaling vs Inference Time Scaling

1. Training Time Scaling:

Training time scaling increases model intelligence by spending more computation upfront on data and network size during initial development.

2. Inference Time Scaling:

Inference time scaling increases the model accuracy by spending more computation at runtime to let the model generate hidden reasoning steps

```python
# Tricky User Prompt

easy_puzzle = [
    {"role": "user", "content": 
        "You toss 2 coins. One of them is heads. What's the probability the other is tails? Answer with the probability only."},
]
```

```python
response = gemini.chat.completions.create(
    model = 'gemini-3.5-flash-lite',
    messages=easy_puzzle,
    reasoning_effort='minimal'                  # tells the model to use least amount of extra reasoning [faster response, less computation, good for simple tasks]
)                                               # reasoning_effort values could be 'minimal', 'low', 'medium' and 'high'

display(Markdown(response.choices[0].message.content))

# In this instance it gave correct result, but in some cases it might give incorrect result due to the nature of focusing on speed rather than accuracy
```

```python
# Inference Time Scaling - using the same model with increased reasoning effort in runtime

response = gemini.chat.completions.create(
    model = 'gemini-3.5-flash-lite',
    messages=easy_puzzle,
    reasoning_effort='low'
)

display(Markdown(response.choices[0].message.content))
```

```python
# Training Time Scaling - using a different model [trained on huge data compared to previous model] with same reasoning effort

response = gemini.chat.completions.create(
    model = 'gemini-3.5-flash',
    messages=easy_puzzle,
    reasoning_effort='minimal'
)

display(Markdown(response.choices[0].message.content))
```

#### Note

We can clearly observe the difference in the runtime: 

- as we increase the reasoning effort, model spent longer time in reasoning to give the better output

- as we change the model from smaller [lite] version to bigger version with same reasoning effort, it spent even more time as the model itself is trained on huge dataset to solve more complex problems

- lite versions are made for fast response and easy tasks 

    - tradeoff: accuracy for speed

- normal versions are made for complex tasks

    - tradeoff: speed for complexity

## GEMINI Client Library

```python
# Using GEMINI client library instead of OpenAI client library

from google import genai

client = genai.Client()

response = client.models.generate_content(
    model = 'gemini-3.5-flash-lite',
    contents = "Describe the color Blue to someone who's never been able to see in 1 sentence"  # Unlike OpenAI client library, Gemini client library explicitly designed to accept a raw string as the user prompt directly
)

print(response.text)
```

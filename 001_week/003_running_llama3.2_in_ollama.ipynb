### Approach 1 - llama3.2 using OpenAI Endpoint (Web Architecture)

```python
from openai import OpenAI

OLLAMA_BASE_URL = 'http://localhost:11434/v1'

ollama = OpenAI(base_url=OLLAMA_BASE_URL, api_key='ollama')

response = ollama.chat.completions.create(model='llama3.2',messages=[{'role':'user', 'content':'Tell me about yourself'}])

response.choices[0].message.content
```

### Note

1. Here we are not using any of the OpenAI's models, servers or paid services.

2. Code is running 100% locally

3. We are using OpenAI python library strictly as a translator tool

4. In the world of LLM Engineering, OpenAI's API structure has became an unofficial industry standard

5. Because almost every AI developer already knew how to write OpenAI code, Ollama and other model vendors designed their local engines to mimic OpenAI's web architecture perfectly

6. By importing OpenAI, we can talk to different backends using a unified coding language

### Approach 2 - llama3.2 using HTTP requests and repsone

```python
import requests

headers = {'Authorization': f'Bearer ollama', 'Content-Type':'application/json'}

payload = {
    'model' : 'llama3.2',
    'messages' : [{'role':'user', 'content':'What is so interesting about India?'}],
    # ollama by default streams its response token by token. To create a json response, we must stop streaming
    'stream': False
}

response = requests.post(
    'http://localhost:11434/v1/chat/completions',
    headers=headers,
    json=payload
)

response.json()

```

```python
response.json()['choices'][0]['message']['content']
```

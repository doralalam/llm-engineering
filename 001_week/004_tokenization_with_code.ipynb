```python
import tiktoken

# to create a tokenizer object
encoder = tiktoken.encoding_for_model("gpt-4.1-mini")
# to convert the text string into a list of tokens
tokens = encoder.encode("Hi, my name is Lalam!")
```

```python
print(tokens)
```

```python
# to convert each token into the respective text string
for token_id in tokens:
    text = encoder.decode([token_id])
    print(f"{token_id} - {text}")
```

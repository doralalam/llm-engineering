```python
# To import OpenAI python client library to use OpenAI endpoint API
# To load environment variables from a file called .env
import os
from openai import OpenAI
from dotenv import load_dotenv

# Load environment variables from a file called .env into the process environment (virtula environment)
load_dotenv(override=True)
# Get the environment variable (OPENAI_API_KEY) from the process environment
api_key = os.getenv('OPENAI_API_KEY')

# Check the key
if not api_key:
    print("No API key was found - please head over to the troubleshooting notebook in this folder to identify & fix!")
elif not api_key.startswith("sk-proj-"):
    print("An API key was found, but it doesn't start sk-proj-; please check you're using the right key - see troubleshooting notebook")
elif api_key.strip() != api_key:
    print("An API key was found, but it looks like it might have space or tab characters at the start or end - please remove them - see troubleshooting notebook")
else:
    print("API key found and looks good so far!")


message = 'Hello GPT! I am new to learn about LLM'

messages = [{'role': 'user', 'content': message}]

messages
```

## Approach 1

1. Using raw HTTP request

```python
import requests

headers = {"Authorization": f"Bearer {api_key}", "Content-Type": "application/json"}

payload = {
    "model": "gpt-40-mini",
    "messages": messages
}

print(payload)

# To send a request to the OpenAI API
response = requests.post(
    "https://api.openai.com/v1/chat/completions",
    headers=headers,
    json=payload
)

response.json()

# We will get error since we are using free tier.
# We need to pay to use OpenAI models
```

```python
# To get only the content of first choise from the json response
response.json()['choices'][0]['message']['content']
```

```python
# To get the content of the first choice without the response.json() we use
response.choices[0].message.content
```

## Approach 2

1. Using OpenAI endpoint

```python
# To create a client object for OpenAI
openai = OpenAI()   # By default, OpenAI() looks for OPENAI_API_KEY

# To create a chat completion
response = openai.chat.completions.create(model="gpt-4o-mini", messages=messages)

# To get the first choice
response.choices[0].message.content

# We got the error since we are using the free tier of OpenAI.
# To fix this, we need to use the paid tier of OpenAI.
```

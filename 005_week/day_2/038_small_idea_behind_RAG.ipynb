#### Different Inference Techniques

- We are already using different inference techniques in-order to get the best and more accurate output

- Inference techniques we are already using such as:

    1. Multi-Shot Prompting:

        - Giving multiple examples to the model before asking for an answer to a particular question

    2. Use of Tools:

        - Using multiple tools along with the models to perform actions that an LLM might not do alone

        - Example: The model doesn't run the IDE directly; instead, it outputs structured data (like JSON or code blocks) requesting a tool execution. An external runtime (your Python script or agent framework passed as a tool) executes that code in the environment and passes the stdout back to the model.

    3. Providing Additional Context:

        - We provide the input prompt along with additional context so the ouput will be based on the best interest of the context

        - Passing retrieved documentation or user background reduces hallucinations and grounds the model's outputs in factual data.

#### RAG (Retrieval-Augmented Generation)

1. Retrieval: (Finding Facts)

    - The system searches an external knowledge source (such as a vector database, document store, or search engine) to find relevant chunks of information matching the user's query.

2. Augmentation: (Enriching the Prompt)

    - The system takes the retrieved facts and injects them directly into the user's original input prompt, providing the LLM with up-to-date or private context.

3. Generation: (Producing the Output)

    - The Large Language Model (LLM) reads the augmented prompt and generates a natural language response grounded strictly in the provided context, reducing hallucinations.

#### Small Idea Behind RAG

- The core idea behind RAG is turning a closed-book AI into an open-book test-taker. Instead of forcing an LLM to rely strictly on what it memorized during training, RAG lets it look up real-time, specific facts from your custom documents before writing an answer.

**The 3-Step Flow (An Everyday Analogy)**

1. Ask (Query): You ask the system, "What is our company's refund policy for international customers?"

2. Look Up (Retrieve): Instead of guessing, the system instantly searches your private database or folder for the relevant policy paragraphs.

3. Answer (Augment & Generate): The system pastes those retrieved paragraphs directly into the prompt and tells the LLM: "Read these exact facts and answer the user's question based strictly on them."

**Why It Matters**
1. Fixes Hallucinations: Prevents the AI from making up believable-sounding lies when it doesn't know an answer.

2. Keeps Data Current: Eliminates the need to constantly retrain or fine-tune multi-billion-parameter models just to update facts or company news.

3. Unlocks Private Data: Enables general AI models to safely answer questions using internal documents, enterprise databases, or personal notes.

`In short: Find the exact facts first, then let the AI write the answer.`

#### Big Idea Behind RAG

**LLMs are of 2 types:**

1. Auto-Regressive LLMs:

    - Trained to take an input sequence and predict the next token

    - "Auto" means self, and "regressive" refers to using previous outputs to predict future ones. An autoregressive model is named this because it feeds its own previously generated tokens back into itself as input to predict the very next token, repeating this loop step-by-step.

    - We use Auto-Regressive LLMs to provide the output based on the input prompt from the user and relevant context retreived from the knowledge store with the help of Encoder

    - LLMs we have used so far like gemini-3.6-flash, gpt-oss-20b, llama-3.1-8b all are auto-regressive LLMs and on the decoder only architecture of Transformers

2. Encoders / Embeddinge Models / Vector Embedding Models:

    - Used to calculate `Vector Embeddings`, representing an input as a list of numbers i.e., a vector

    - These LLMs are used in tasks like Sentiment Analysis, Classification etc.,

    - It maps text into an N-dimensional semantic space. Rather than classifying text into fixed categories, it allows vector databases to use distance metrics (like Cosine Similarity or Euclidean Distance) to measure how close two ideas are in meaning.

    - We use Encoders to convert the input tokens into Vectors, which works as the text with similar meaning was always held close in higher dimensions.

    - Modern vector embedding models are encoder only models.

    - Classical Autoencoders contain two parts (an encoder to compress data and a decoder to reconstruct the original input). Embedding models use only the encoder component to map text into a dense vector space without needing to reconstruct the original input.

    - Example for these models are:

        - BERT from Google - Bi-directional Encoder Representations from Transformers

        - OpenAIEmbeddings from OpenAI

        
`Almost all autoregressive generative LLMs use a decoder-only architecture, while dedicated embedding and retrieval models use an encoder-only architecture.`



---

### Key Architectural Differences

| Feature | Decoder-Only (Autoregressive) | Encoder-Only (Embeddings) |
| :--- | :--- | :--- |
| **Primary Job** | **Text Generation** (predicting what comes next). | **Text Understanding** (creating rich numerical vectors). |
| **Attention Mechanism** | **Causal / Masked Self-Attention** (can only look left at previous tokens). | **Bidirectional Self-Attention** (looks left and right across all tokens at once). |
| **Output** | A sequence of newly generated text tokens. | A single fixed-length vector array representing the whole text. |
| **Examples** | Llama 3, GPT-4, Gemini, DeepSeek-R1, Mistral. | BERT, RoBERTa, BGE, E5, Nomic Embed. |

---

### Why the Architectures Are Built Differently

* **Why Decoders use Masked Attention:** 

    - When an autoregressive model generates text, it cannot "cheat" by looking at future tokens that haven't been written yet. 
    
    - Masking hides future tokens so the model learns strictly to predict token $T_{N+1}$ based on tokens $T_1$ through $T_N$.

* **Why Encoders use Bidirectional Attention:** 

    - When creating a vector embedding for a document, the goal isn't generation—it's deep comprehension. The model needs to see the *entire sentence simultaneously* because the meaning of a word depends heavily on words that come both before and after it.

---

### Is there an Encoder-Decoder architecture?

Yes, the original 2017 Transformer paper introduced an **Encoder-Decoder** (or Sequence-to-Sequence) architecture:

* The **Encoder** reads the entire input bidirectionally, and the **Decoder** generates the output autoregressively.

* **Examples:** T5, BART, and Whisper (for speech recognition). 

While popular for translation and summarization, the industry overwhelmingly shifted to **Decoder-Only** models for general LLMs because they scale much more efficiently for open-ended text generation.

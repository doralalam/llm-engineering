#### 10 Advanced RAG Techniques to Improve the RAG Performance

1. **Chunking R&D**

    - Experimenting with Chunking strategies like Semantic chunking

2. **Encoder R&D**

    - Select the best encoder model based on the dataset

    - SentenceTransformer is best for Image+Text

3. **Improve Prompts**

    - General content, the current date, relevant context and history

4. **Document Pre-Processing or Document Re-Writing**

    - Use an LLM to make the chunks and / or text for encoding

    - Example: Ask an LLM to rewrite the knowledge base so it can be best optimized for vector storage to make it most suitable for querying

5. **Query Re-Writing**

    - Use an LLM to convert the user's question into a RAG query

6. **Query Expansion**

    - Use an LLM to turn the question into multiple RAG queries

7. **Re-Ranking**

    - Use an LLM to sub-select the RAG results after Query Expansion

    - After query expansion, we might have n-queries with each query retrieving K chunks

    - So, in total, after the query expansion stage, we might end up having n x K chunks

    - To get the better RAG metrics like MRR, nDCG, Recall@K, we need to re-rank to keep the most important chunks at the top and we use LLM for this task

8. **Hierarchical RAG**

    - With more number of relevant chunks, RAG performance might not be up to mark due to the self-attention mechanism

    - So, to overcome this, we can use LLMs to summarize at multiple levels

    - Example: If there is a table of employees along with salaries, an we need the number of employees earning salary less than x, this is some what difficult for a single LLM

    - To solve this, it is best to use a separate LLM to summarize i.e., to select the employees whose salary less than x and provide this summary to another LLM which can use this to answer our question by counting

9. **Graph RAG**

    - Retrieve content closely related to similar documents

10. **Agentic RAG**

    - Use Agents for retrieval, combining with memory and tools such as SQL

```python
# Now import directly from evaluation

from evaluation import test
```

```python
# load the test cases from tests.jsonl via test.py
# jsonl - stands for JSON lines i.e., line-delimited JSON objects
# in jsonl file, each line is a json object instead of the entire file as a json object

tests = test.load_tests()
```

```python
# count the test cases

len(tests)
```

```python
# sample test case

example = tests[0]

print(example.question)                  # since we are accessing jsonl objects, we have to use example.question instead of example['question']
print(example.keywords)
print(example.reference_answer)
print(example.category)
```

```python
# to count the individual varieties

from collections import Counter

count = Counter([t.category for t in tests])
count

# so, there are a total of 150 test cases with 7 different categories
```

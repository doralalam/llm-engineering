#### Pro's and Con's of RAG along with Evaluation Metrics for Retrieval and Answer

- `Note` : 

    - Chunking strategy will have high impact on performance of the RAG

#### Pro's of RAG

- Rapid time to market

- Give an LLM vast experience

- Save on the context length

- Avoid distracting with irrelevant context

#### Con's of RAG

- RAG is just a hack

    - In RAG, a small context was fed to the transformer and the output depends on the context 

    - If the context is not appropriate then the result is not appropriate as well

- RAG is very much experimental, more of like a trial and error

#### Evaluations are everything

-  Curate a Test Set:

    - Example questions set with the right context identified and reference answers provided

- Measure Retrieval:

    - MRR : Mean Reciprocal Rank

    - nDCG: Normalized - Discounted Cumulative Gain

    - Recall@K

    - Precision@K

- Measure Answer:

    - Use LLM as a judge to score provided answers against criteria like:

        - Accuracy

        - Completeness
        
        - Relevance



#### 1 Measure Retrieval (as part of RAG evaluation)

1. **MRR - Mean Reciprocal Rank**

    - Average inverse Rank of first hit:

    - Reciprocal Rank for query = 1, if the first chunk always has relevant chunk

    - Reciprocal Rank for query = 1/2, if the second chunk has the relevant chunk

    - Reciprocal Rank for query = 1/3, if the third chunk has the relevant chunk ........

    - MRR = Average(Reciprocal Ranks of all queries or the dataset)

2. **nDCG -  Normalized Discounted Cumulative Gain**

    - Measures how well a retriever ranks search results based on both relevance grade and position

    - It answers: Are the most relevant items placed at the very top of the list? If the relevant chunks are at the bottom, it penalizes to ensure the most relevant chunks to be at the top

- Core Mechanics:

    1. Graded Relevance: 
    
        - Unlike binary metrics (relevant vs. irrelevant), nDCG assigns scale ratings (e.g., 0 = useless, 1 = weak, 2 = relevant, 3 = perfect).

    2. Logarithmic Discounting: 
    
        - Items lower in the list receive a position penalty. A highly relevant item at rank 1 produces a high score; at rank 8, its contribution is heavily discounted.

    3. Normalization (0.0 to 1.0): 
    
        - The raw score (DCG) is divided by the maximum possible score of a perfect ordering (IDCG).

                `nDCG = (DCG/IDCG)`
    ​

3. **Precision@K**

    - Measures `accuracy` - what proportion of the top K retrieved items are actually relevant

    `Precision@K = (Number of relevant items in the top K) / K`

4. **Recall@K**

    - Measures `keyword coverage` - what proportion of all the relevant items in the dataset  are retrieved in the top K

    `Recall@K = (Number of relevant items in the top-K / Total number of relevant items in the dataset (groung truth))` 



#### 2 Measure Answer (as part of RAG evaluation)

- Use an LLM as a Judge to measure

    1. **Accuracy** (Hallucination Rate)

        - Measures whether every statement in the generated answer is factually correct and strictly grounded in the retrieved context.

        - `Accuracy Score = (Number of claims grounded in the context) / (Total claims in the answer)`

    2. **Completeness** (Coverage)

        - Measures whether the generated answer addresses every part of the user's prompt and captures all critical information present in the source material.

        - `Completeness Score = (Key Information points covered in the answer) / (Total key information points required)`

    3. **Relevance** (Answer Relevance)

        - Measures how directly and concisely the answer addresses the original query, without introducing unnecessary or off-topic information.

        - `Relevance Score = Cosine Similarity(Original Query,Predicted Query from Answer)`

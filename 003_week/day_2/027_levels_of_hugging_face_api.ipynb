## **The Two API Levels of Hugging Face**

1. **Pipelines**

  - Higher level APIs to carry out standard tasks incredibly quickly

  - Pipelines are for the simple out-of-the box inference tasks
  
  - Useful for simple and common use cases like Sentiment Analysis, classifier, Named Entity Recognition, Question Answering, Summarization, Translation etc.,


2. **Tokenizers and Models**

  - Lower level APIs to provide the most power and control

  - These are the real granular APIs where we get to work with the underlying deep neural network and gives us the most power and control

**In Simple**

1. Pipelines are quick and simple, whereas

2. Tokenizers and Models are where we get the raw power

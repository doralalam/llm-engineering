### HuggingFace

- HuggingFace is like a github but for data science models and also for datasets

- Before HuggingFace, Kaggle is the place for datasets

- HuggingFace consists 3 main components:

    1. Data Science Models

    2. Datasets

    3. Spaces

- Spaces are like hosted apps where we host a software to make it visible to other users


### HuggingFace Libraries

- HuggingFace has the open source libraries that implement open source transformer models using Pytorch (most popular) or TensorFlow or Jax which are different deep neural networks

- These transformer libraries were written in Pytorch so that we can import them using Python

#### Difference between Ollama and HuggingFace Libraries

##### Ollama

- Ollama is like a software product which packages all the requirements for models like Gemma or llama into GIF file and runs with efficients C++ code locally in a box and it listens in a particular port which it provides an endpoint that is compatible with OpenAI.

- So we can use OpenAI's libraries to connect to it. It is like a fast packaged pre-baked software

- Finally, in ollama, we won't get the actual code for the transformers which were written in Pytorch. We just connect to the endpoint and use

##### HuggingFace Libraries

- In HuggingFace libraries, we get to access to the actual code and we can step through that code and run the model ourself

- With this we can mess with the code such as tokens or swap out some layers of the neural network or can do fine-tuning which none of these are possible in ollama

- HuggingFace offers a set of open source libraries for running Python code to run neural networks with transformers as part of it






### Six different HuggingFace libraries

1. Hub:

    - A Python library let's us to connect with the HuggingFace hub to download the models, datasets that are already available in the HuggingFace website

2. Datasets:

    - A python library to import the datasets that already available in the HuggingFace website

3. Transformers:

    - A python library that let's us to import the code for a particular model and run that code or train it

4. PEFT: Paramater Efficient Fine Tuning

    - A python library used to train the model without tweeking the actual weights using techniques like LoRa (Low Rank Adaptation) or QLoRa (Quantized Low Rank Adaptation)

5. TRL: Transformers Reinforcement Learning

    - A python library used to train the transformers

6. Accelerate:

    - A python library used to distribute models across multiple GPUs

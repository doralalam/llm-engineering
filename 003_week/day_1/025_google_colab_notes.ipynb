### Google Colab

- Google Colab allows us to run a Jupyter notebook in the cloud with a powerful runtime

- It allows us to collaborate with others via sharing

- It allow us to integrate with other Google services

- **Big Deal: It allow us to access some high end GPUs**

#### Runtimes that are available in Google Colab

1. CPU based runtime

2. Lower spec GPU for free or low-cost T4 (15GB GPU RAM)

3. Higher spec GPU for resource intensive runs - A100 (40GB)
